```python
def getBondPrice(y, face, couponRate, m, ppy=1):

    bondPrice = 0

    for t in range(1, m * ppy + 1):

        cf = face * couponRate / ppy

        pv = cf / (1 + y / ppy) ** t

        bondPrice = bondPrice + pv

    pv = face / (1 + y / ppy) ** (m * ppy)

    bondPrice = bondPrice + pv

    return(bondPrice)
```

```python
y = 0.03
face = 2000000
couponRate = 0.04
m = 10

print(getBondPrice(y, face, couponRate, m))
print(getBondPrice(y, face, couponRate, m, 2))
```

```python
def getBondDuration(y, face, couponRate, m, ppy=1):

    bondPrice = 0
    total = 0

    for t in range(1, m * ppy + 1):
        cf = face * couponRate / ppy

        if t == m * ppy:
            cf = cf + face

        pv = cf / (1 + y / ppy) ** t

        bondPrice = bondPrice + pv
        total = total + (t / ppy) * pv

    bondDuration = total / bondPrice

    return(bondDuration)
```

```python
y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1

getBondDuration(y, face, couponRate, m, ppy)
```

```python
def getBondPrice_E(face, couponRate, m, yc):

    bondPrice = 0

    for i, y in enumerate(yc):

        t = i + 1

        cf = face * couponRate

        if t == m:
            cf = cf + face

        pv = cf / (1 + y) ** t

        bondPrice = bondPrice + pv

    return(bondPrice)
```

```python
yc = [0.010, 0.015, 0.020, 0.025, 0.030]
face = 2000000
couponRate = 0.04
m = 5

getBondPrice_E(face, couponRate, m, yc)
```

```python
def getBondPrice_Z(face, couponRate, times, yc):

    bondPrice = 0

    for t, y in zip(times, yc):

        cf = face * couponRate

        if t == times[-1]:
            cf = cf + face

        pv = cf / (1 + y) ** t

        bondPrice = bondPrice + pv

    return(bondPrice)
```

```python
yc = [0.010, 0.015, 0.020, 0.025, 0.030]
times = [1, 1.5, 3, 4, 7]
face = 2000000
couponRate = 0.04

getBondPrice_Z(face, couponRate, times, yc)
```

```python
def FizzBuzz(start, finish):

    outlist = []

    for i in range(start, finish + 1):

        if i % 15 == 0:
            outlist.append("fizzbuzz")

        elif i % 3 == 0:
            outlist.append("fizz")

        elif i % 5 == 0:
            outlist.append("buzz")

        else:
            outlist.append(i)

    return(outlist)
```

```python
FizzBuzz(1, 15)
```
