```python
import numpy as np

def getBondPrice(y, face, couponRate, m, ppy=1):
    periods = np.arange(1, m * ppy + 1)
    coupon = face * couponRate / ppy
    pv = 1 / (1 + y / ppy) ** periods
    coupon_pv = coupon * pv
    price = np.sum(coupon_pv) + face * pv[-1]
    return price
```

```python
y = 0.03
face = 2000000
couponRate = 0.04
m = 10

print(getBondPrice(y, face, couponRate, m))
print(getBondPrice(y, face, couponRate, m, 2))
```

```python
import numpy as np

def getBondDuration(y, face, couponRate, m, ppy=1):
    periods = np.arange(1, m * ppy + 1)
    pv = 1 / (1 + y / ppy) ** periods

    cf = np.ones(m * ppy) * face * couponRate / ppy
    cf[-1] = cf[-1] + face

    pvcf = cf * pv
    price = np.sum(pvcf)

    duration = np.sum(periods / ppy * pvcf) / price

    return duration
```

```python
y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1

print(getBondDuration(y, face, couponRate, m, ppy))
```
